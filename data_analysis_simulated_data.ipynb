```python
import numpy as np
import pandas as pd
import os
```

```python
import_folder_path = './data/simulated_data'
keyword = 'simulated'

csv_files = [f for f in os.listdir(import_folder_path) if 'csv' in f and keyword in f]
csv_files
```

# Calculate episode parameters

```python
n_rows = 0
dfs = []
for f in csv_files:
    df = pd.read_csv(import_folder_path + '/' + f)
    n_rows += len(df)

    # Step 1. Calculate  number and duration of duration episodes 
    # Add a column to identify consecutive groups
    df['group'] = (df['detected'] != df['detected'].shift()).cumsum()

    # Filter detected groups and calculate counts
    results = (
        df[df['detected']]  # Filter detected rows
        .groupby('group')   # Group by consecutive detected groups
        .size()             # Get the size of each group
        .reset_index(name='count')  # Rename the count column
    )

    results['handflapping_episode'] = range(1, len(results) + 1)
    results['duration'] = results['count'] / 50
    results['csv_file'] = f

    dfs.append(results)
```

```python
merged_df = pd.concat(dfs, ignore_index=True)
merged_df
```

```python
# Calculate and print the mean and standard deviation of durations
mean_duration = merged_df['duration'].mean()
std_duration = merged_df['duration'].std()

print(f'The mean duration of episodes is: {mean_duration:.2f} seconds')
print(f'The standard deviation of durations is: {std_duration:.2f} seconds')
```

```python
# Calculate and print the mean and standard deviation of episode frequency
episode_frequency = merged_df.groupby('csv_file').size().reset_index(name='count') 
episode_frequency['frequency'] = episode_frequency['count'] / 2

mean_frequency = episode_frequency['frequency'].mean()
std_frequency = episode_frequency['frequency'].std()

print(f'The mean frequency of episodes is: {mean_frequency:.2f} seconds')
print(f'The standard deviation of frequency is: {std_frequency:.2f} seconds')
```
